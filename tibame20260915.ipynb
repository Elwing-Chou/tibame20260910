```python
# 我個人會選這個, 因為這個比較好擴充
# 快捷鍵: TAB Shift+TAB
def is_leap_year(year):
    # year = 2000
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False

print(is_leap_year(2000))
print(is_leap_year(1800))
```

```python
# 複習: 定義功能
# 參數: 代號
# 回傳值: return
def add(n1, n2):
    n3 = n1 + n2
    return n3

print(add(3, 5))
print(add(4, 6))
```
