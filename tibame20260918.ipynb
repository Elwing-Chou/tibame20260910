```python
# oxoxo
# xoxox
# oxoxo
# xoxox
# TODO: 如果有三個以上符號輪迴(list)
for i in range(4):
    line = ""
    for j in range(5):
        if (j + i) % 3 == 0:
            line = line + "o"
        elif (j + i) % 3 == 1:
            line = line + "x"
        elif (j + i) % 3 == 2:
            line = line + "!"
    print(line)
```

```python
symbols = ["o", "x", "!"]
for i in range(4):
    line = ""
    for j in range(5):
        pos = (j + i) % 3
        line = line + symbols[pos]
    print(line)
```
