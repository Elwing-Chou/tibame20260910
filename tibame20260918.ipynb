```python
# oxoxo
# xoxox
# oxoxo
# xoxox
# TODO: 如果有三個以上符號輪迴(list)
for i in range(4):
    line = ""
    for j in range(5):
        if (j + i) % 3 == 0:
            line = line + "o"
        elif (j + i) % 3 == 1:
            line = line + "x"
        elif (j + i) % 3 == 2:
            line = line + "!"
    print(line)
```

```python
symbols = ["o", "x", "!", "u"]
for i in range(4):
    line = ""
    for j in range(5):
        pos = (j + i) % len(symbols)
        line = line + symbols[pos]
    print(line)
```

```python
# 初始化一個list
# 方法1.
a = [0] * 3
print(a)
# 方法2. 初始化, 順手填值(!)
a = []
for i in range(5):
    a.append(i)
print(a)
# 方法3. 把方法2精簡
a = [(i + 1) for i in range(5)]
print(a)
```

```python
# 二維list的初始化
# 錯的, 因為你b只創造了一次, 你就算裡面有三個, 還是指同一個
b = [0] * 5
a = [b] * 3
print(a)
a[0][0] = 99
print(a)
# 方法2
a = []
for i in range(3):
    b = [0] * 5
    a.append(b)
a[0][0] = 88
print(a)
# 方法3
a = [[0] * 5 for i in range(3)]
a[0][0] = 77
print(a)
```

```python
# a = [
#     [0, 1, 2, 3, 4],
#     [9, 8, 7, 6, 5],
#     [10, 11, 12, 13, 14],
#     [19, 18, 17, 16, 15]
# ]
row, col = 4, 5
# a = [[0] * col for i in range(row)]
a = []
for i in range(row):
    b = [0] * col
    a.append(b)

start = 5
for i in range(row):
    for j in range(col):
        if i % 2 == 0:
            a[i][j] = col * i + j + start
        else:
            a[i][j] = (col * i + (col - 1)) - j + start
print(a)
```

- 四大群集
- list: []
    - 清單, 同類型的很多東西
    - key: 0 1 2 3
    - 有序

- dict: {}
    - 字典, 聚集很多欄位, 變成一個複雜單位
    - key是自訂的, "name":"Elwing"
    - 無序

- set: {}
    - 集合, 裡面元素不重複的list
    - key是python內部自己訂的, 你不能用
    - 無序

```python
s1 = {20, 30, 10, 20}
print(s1)
# 交集: 共同有的(&)
# 聯集: 取所有的(|)
# 減集: 減掉你的(-)
s2 = {20, 5}
print(s1 & s2)
print(s1 | s2)
print(s1 - s2)
# 增加東西: add
# 減少東西: discard
s1.add(40)
s1.add(20)
print(s1)
s1.discard(4)
s1.discard(20)
print(s1)
# for...in
for c in s1:
    print(c)
```
