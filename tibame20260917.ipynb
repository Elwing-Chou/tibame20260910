```python
# 我個人會選這個, 因為這個比較好擴充
# 快捷鍵: TAB Shift+TAB
def is_leap_year(year):
    # year = 2000
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False

print(is_leap_year(2000))
print(is_leap_year(1800))
```

```python
def is_date_valid(d):
    # d = 20240229
    day = d % 100
    remain = d // 100
    month = remain % 100
    year = remain // 100
    # print(year, month, day)

    if year >= 1:
        if 12 >= month >= 1:
            day_per_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
            max_days = day_per_month[month-1]
            # if month == 2 and is_leap_year(year) == True
            if month == 2 and is_leap_year(year):
                max_days = max_days + 1
            if max_days >= day >= 1:
                return True
            else:
                # day invalid
                return False
        else:
            # month invalid
            return False
    else:
        # year invalid
        return False

print(is_date_valid(20240229))
print(is_date_valid(20240230))
```

```python
def thought_2(start=4499, maxcount=10):
    # start = 4499
    count = 0
    result = []
    while True:
        d = start * start
        # if is_date_valid(d) == True
        if is_date_valid(d):
            # print(d)
            result.append(d)
            count = count + 1
        if count == maxcount:
            break
        start = start + 1
    return result

print(thought_2())
print(thought_2(maxcount=2))
print(thought_2(start=1, maxcount=5))
```

```python
# python的日期時間模組有兩個
# 1. time: 時間
# 2. datetime: 日期+時間
# str<->datetime 型態轉換
# strptime: str -> datetime
# strftime: datetime -> str
import datetime as dt
s = "2026-09-15"
d = dt.datetime.strptime(s, "%Y-%m-%d")
print(d)

# 專屬功能
s2 = d.strftime("%m/%d(%a)")
print(s2)
```

```python
# 想法1. 迴圈
import datetime as dt
def thought_1(y=2024, m=10, d=1, maxcount=10):
    d = dt.datetime(y, m, d)
    delta = dt.timedelta(days=1)
    count = 0
    result = []
    while True:
        # 一天往後數
        # 開根號(**0.5)->確認是否是整數
        d_str = d.strftime("%Y%m%d")
        d_int = int(d_str)
        d_root = d_int ** 0.5
        if d_root.is_integer():
            # 如果是整數, 就是個完全平方日
            # print(d_int)
            result.append(d_int)
            count = count + 1
            if count == maxcount:
                break
        d = d + delta
    return result

print(thought_1(y=7777, m=10, maxcount=5))
```
